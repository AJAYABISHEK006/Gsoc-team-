```python
song = song.fillna({
    "genre": "Unknown",
    "artist_name": "Unknown",
    "track_name": "Unknown",
    "artists": "Unknown",
    "album_name": "Unknown"
})
```

```python
features = [
    'danceability',
    'energy',
    'loudness',
    'speechiness',
    'acousticness',
    'instrumentalness',
    'liveness',
    'valence',
    'tempo'
]

X = song[features]
```

```python
X.head()
```

```python
X.shape
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
X_scaled = scaler.fit_transform(X)
```

```python
from sklearn.neighbors import NearestNeighbors
```

```python
model = NearestNeighbors(
    n_neighbors = 11,
    metric = "cosine"
)
```

```python
model.fit(X_scaled)
```

```python
def recommend(song_name, n=10):

    matches = song[song["track_name"].str.lower() == song_name.lower()]

    if matches.empty:
        print("Song not found")
        return

    song_index = matches.index[0]

    print("Selected song:", song.loc[song_index, "track_name"])
    print("Artist:", song.loc[song_index, "artists"])
    print("\nRecommended songs:\n")

    distances, indices = model.kneighbors(
        X_scaled[song_index].reshape(1, -1),
        n_neighbors=n + 1
    )

    count = 0

    for i in indices[0]:

        if i == song_index:
            continue

        print(
            f"{count + 1}. "
            f"{song.loc[i, 'track_name']} - "
            f"{song.loc[i, 'artists']}"
        )

        count += 1

        if count == n:
            break
```

```python
recommend("Comedy")
```
