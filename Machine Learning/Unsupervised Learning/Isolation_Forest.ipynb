```python
import  numpy as np
import matplotlib.pyplot as plt
from sklearn.ensemble import IsolationForest
```

```python
x = np.array([
    [10],[11],[12],[13],[14],[15],[16],[17],[50],[18],[20]
])
```

```python
model = IsolationForest(
    contamination = 0.1,
    random_state=42
)
```

```python
labels = model.fit_predict(x)
```

```python
print("Labels: ")
print(labels)
```

```python
for value, label in zip(x,labels):
  if label == -1:
    print(value[0], "-> Anomaly")
  else:
    print(value[0],"-> Normal")
```
