```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
```

```python
iris = load_iris()
```

```python
x = pd.DataFrame(
    iris.data,
    columns = iris.feature_names
)
```

```python
kmeans = KMeans(
    n_clusters = 3,
    random_state = 42
)
```

```python
kmeans.fit(x)
```

```python
x["Cluster"] = kmeans.labels_
```
