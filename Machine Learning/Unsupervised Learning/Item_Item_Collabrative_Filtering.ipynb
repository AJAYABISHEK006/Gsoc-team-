```python
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity
```

```python
data = {
    "User": [
        "Alice", "Alice", "Alice",
        "Bob", "Bob", "Bob",
        "Charlie", "Charlie", "Charlie"
    ],

    "Movie": [
        "Inception",
        "Interstellar",
        "Titanic",
        "Inception",
        "Interstellar",
        "Avengers",
        "Titanic",
        "Avengers",
        "Inception"
    ],

    "Rating": [
        5, 5, 1,
        5, 4, 5,
        5, 4, 1
    ]
}
```

```python
df = pd.DataFrame(data)
```

```python
print("Original Data")
print()
print(df)
```

```python
user_movie_matrix = df.pivot_table(
    index = "User",
    columns = "Movie",
    values = "Rating"
)
```

```python
print("\nUser-Movie Matrix")
print()
print(user_movie_matrix)
```

```python
user_movie_matrix = user_movie_matrix.fillna(0)
```

```python
print(user_movie_matrix)
```

```python
movie_user_matrix = user_movie_matrix.T
```

```python
print(movie_user_matrix)
```

```python
movie_similarity = cosine_similarity(movie_user_matrix)
```

```python
print(movie_similarity)
```

```python
movie_similarity_df = pd.DataFrame(
    movie_similarity,
    index = movie_user_matrix.index,
    columns = movie_user_matrix.index
)
```

```python
print(movie_similarity_df)
```

```python
def recommend(movie):

  if movie not in movie_user_matrix.index:
    print("Movie not found!")
    return []

  movie_index = movie_user_matrix.index.get_loc(movie)

  scores = list(enumerate(movie_similarity[movie_index]))

  scores = sorted(scores, key = lambda x: x[1], reverse=True)

  recommendations = []

  for index, score in scores[1:4]:

    movie_name = movie_user_matrix.index[index]

    similarity_score = round(score*100,2)

    recommendations.append((movie_name, similarity_score))

  return recommendations
```

```python
recommendations = recommend("Inception")
```

```python
for movie, score in recommendations:

    print(movie,"-",score,"%")
```
