```python
import pandas as pd
from mlxtend.frequent_patterns import fpgrowth, association_rules
```

```python
data = {
    'Milk': [1,1,1,0,1],
    'Bread': [1,1,0,1,1],
    'Butter': [1,0,1,1,1]
}
```

```python
df = pd.DataFrame(data)
print(df)
```

```python
frequent_itemsets = fpgrowth(
    df,
    min_support = 0.6,
    use_colnames = True
)

print(frequent_itemsets)
```

```python
rules = association_rules(
    frequent_itemsets,
    metric = "confidence",
    min_threshold = 0.7
)

print(rules)
```
