```python
import pandas as pd
```

```python
data = {
    "User": [
        "Alice",
        "Alice",
        "Alice",
        "Bob",
        "Bob",
        "Bob",
        "Charlie",
        "Charlie",
        "Charlie"
    ],

    "Movie": [
        "Inception",
        "Interstellar",
        "Titanic",
        "Inception",
        "Interstellar",
        "Avengers",
        "Titanic",
        "Avengers",
        "Inception"
    ],

    "Rating": [
        5,
        5,
        1,
        5,
        4,
        5,
        5,
        4,
        1
    ]

}
```

```python
df = pd.DataFrame(data)
```

```python
print(df)
```

```python
user_movie_matrix = df.pivot_table(
    index = "User",
    columns = "Movie",
    values = "Rating"
)
```

```python
print(user_movie_matrix)
```

```python
user_movie_matrix = user_movie_matrix.fillna(0)
```

```python
print(user_movie_matrix)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
user_similarity = cosine_similarity(user_movie_matrix)
```

```python
print(user_similarity)
```

```python
user_index = user_movie_matrix.index.get_loc("Alice")
```

```python
print(user_index)
```

```python
similar_users = list(enumerate(user_similarity[user_index]))
```

```python
print(similar_users)
```

```python
similar_users = sorted(
    similar_users,
    key = lambda x: x[1],
    reverse = True
)
```

```python
print(similar_users)
```

```python
similar_users = similar_users[1:]
```

```python
print(similar_users)
```

```python
print("\nUsers similar to Alice:")

for index, score in similar_users:

  username = user_movie_matrix.index[index]
  print(username, "-", round(score,2))
```

```python
most_similar_user_index = similar_users[0][0]
```

```python
print(most_similar_user_index)
```

```python
most_similar_user = user_movie_matrix.index[most_similar_user_index]
```

```python
print(most_similar_user)
```

```python
alice_ratings = user_movie_matrix.loc["Alice"]
```

```python
print(alice_ratings)
```

```python
similar_user_ratings = user_movie_matrix.loc[most_similar_user]
```

```python
print(similar_user_ratings)
```

```python
recommendations = []
```

```python
for movie in user_movie_matrix.columns:

    similar_rating = similar_user_ratings[movie]

    alice_rating = alice_ratings[movie]

    if similar_rating >= 4 and alice_rating == 0:

        recommendations.append(movie)
```

```python
for movie in recommendations:
  print(movie)
```
