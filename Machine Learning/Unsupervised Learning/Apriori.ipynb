```python
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules
```

```python
data = {
    'Milk': [1,1,0,1,1],
    'Bread': [1,1,1,1,0],
    'Butter': [0,1,1,1,1],
    'Eggs': [1,0,1,0,1]
}
```

```python
df = pd.DataFrame(data)
```

```python
print(df)
```

```python
frequent_itemsets = apriori(
    df,
    min_support = 0.4,
    use_colnames = True
)
```

```python
print(frequent_itemsets)
```

```python
rules = association_rules(
    frequent_itemsets,
    metric = "confidence",
    min_threshold = 0.6
)
```

```python
print(rules)
```

```python
print("\nAssociation Rules:")
print(rules[['antecedents',
             'consequents',
             'support',
             'confidence',
             'lift']])
```
