```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
```

```python
x,y = make_moons(
    n_samples = 300,
    noise = 0.08,
    random_state = 42
)
```

```python
plt.scatter(x[:,0], x[:,1])
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("Original Data")
plt.show()
```

```python
!pip install hdbscan
```

```python
import hdbscan

clusterer = hdbscan.HDBSCAN(
    min_cluster_size = 10,
    min_samples = 5
)

labels = clusterer.fit_predict(x)
print(labels)
```

```python
plt.scatter(
    x[:,0],
    x[:,1],
    c = labels
)

plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("HDBSCAN Clustering")
plt.show()
```
