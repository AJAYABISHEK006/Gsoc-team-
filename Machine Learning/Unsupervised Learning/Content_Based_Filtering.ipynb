```python
import pandas as pd
```

```python
data = {
    "title" : [
        "Inception",
        "Intersteller",
        "The Matrix",
        "Titanic",
        "Avengers"
    ],
    "description" : [
        "A thief enters dreams and steals secrets.",
        "Scientists travel through space to find a new planet.",
        "A hacker discovers a virtual reality world.",
        "A romantic story about a couple on a ship.",
        "Superheroes fight to save the world."
    ]
}
```

```python
df = pd.DataFrame(data)
```

```python
print(df)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
vectorizer = TfidfVectorizer()
```

```python
tfidf_matrix = vectorizer.fit_transform(df["description"])
```

```python
print(tfidf_matrix)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(tfidf_matrix)
```

```python
print(similarity)
```

```python
def recommend(movie):

  index = df[df["title"] == movie].index[0]
  scores = list(enumerate(similarity[index]))
  scores = sorted(scores, key=lambda x: x[1], reverse = True)

  for i in scores[1:3]:
    print(df.iloc[i[0]]["title"])
```

```python
recommend("Inception")
```
