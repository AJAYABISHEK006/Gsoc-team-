```python
import numpy as np
from sklearn.covariance import EllipticEnvelope
```

```python
x = np.array([
    [1,1],
    [2,2],
    [1,2],
    [2,1],
    [1.5,1.5],
    [2,2.5],
    [8,8]
])
```

```python
model = EllipticEnvelope(contamination = 0.1)
```

```python
predictions = model.fit_predict(x)
```

```python
for point, prediction in zip(x, predictions):
  if prediction == 1:
    print(point, "-> Normal")
  else:
    print(point, "-> Anomaly")
```
