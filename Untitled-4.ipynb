```python
para1=set(input("Enter Paragraph 1: ").lower().split())
para2=set(input("Enter Paragraph 2: ").lower().split())
print("Unique Words (Union):",sorted(para1|para2))
print("Common Words (Intersection):",sorted(para1&para2))
print("Words Only in Paragraph 1:",sorted(para1-para2))
print("Words Only in Paragraph 2:",sorted(para2-para1))
print("Words in Exactly One Paragraph (Symmetric Difference):",sorted(para1^para2))
```
